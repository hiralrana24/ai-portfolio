```python
import fitz  # pymupdf

doc = fitz.open("invoice.pdf")
text = ""
for page in doc:
    text += page.get_text()

print(text)
```

```python
print(f"Number of characters extracted: {len(text)}")
print(f"Number of pages: {len(doc)}")
```
