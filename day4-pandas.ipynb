```python
# Pandas - the most important library for your projects !
import pandas as pd

# Create a DataFrame from your invoices
data = {
    "supplier": ["Food & Beverage Co", "Cleaning Services", "Laundry Co", "Food & Beverage Co"],
    "amount": [1250.50, 340.00, 89.50, 2100.00],
    "date": ["2024-01-15", "2024-01-16", "2024-01-17", "2024-01-18"],
    "category": ["Restaurant", "Cleaning", "Laundry", "Restaurant"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Exploring the DataFrame
print(df.shape)        # number of rows and columns
print(df.columns)      # column names
print(df.describe())   # statistics
```

```python
# Filtering data - very useful for your project !
restaurant_invoices = df[df["category"] == "Restaurant"]
print(restaurant_invoices)

print("\n--- Invoices above 500€ ---")
expensive = df[df["amount"] > 500]
print(expensive)
```

```python
# Group by category - sum amounts per category
summary = df.groupby("category")["amount"].sum()
print(summary)
```

```python
# Export to Excel - exactly what your project will do !
df.to_excel("invoices_report.xlsx", index=False)
print("File saved successfully!")
```
