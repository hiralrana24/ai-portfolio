```python
# Les fonctions
def calculate_average(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    return average

my_grades = [15, 18, 12, 17, 14]
result = calculate_average(my_grades)
print(f"My average is: {result}")
```
