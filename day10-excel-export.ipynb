```python
import pandas as pd
from openpyxl import load_workbook

# The data extracted by Mistral yesterday
invoice_data = {
    "invoice_number": "1253888565",
    "date": "27.04.2026",
    "supplier": "SYSCO France SAS",
    "client_reference": "354220",
    "total_ht": 526.47,
    "total_tva": 28.96,
    "total_ttc": 555.43
}

# Convert to DataFrame
df = pd.DataFrame([invoice_data])
print(df)
```
