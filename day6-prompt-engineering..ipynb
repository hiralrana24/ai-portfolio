````python
response = client.chat.complete(
    model="mistral-small-latest",
    messages=[
        {"role": "system", "content": "Extract invoice data and respond ONLY in JSON format like this example: {\"supplier\": \"...\", \"amount\": 0.0, \"date\": \"YYYY-MM-DD\", \"category\": \"...\"}"},
        {"role": "user", "content": "Invoice from Cleaning Services, 340€, January 16 2024, cleaning category"}
    ]
)

print(response.choices[0].message.content)
````

````python
# Clean the response to remove markdown formatting
import json

raw_response = response.choices[0].message.content
clean_response = raw_response.replace("```json", "").replace("```", "").strip()

data = json.loads(clean_response)
print(data)
print(data["amount"])
````
