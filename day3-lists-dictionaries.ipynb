```python
# Lists in Python
fruits = ["apple", "banana", "orange", "mango"]

print(fruits)
print(fruits[0])   # first element
print(fruits[-1])  # last element
print(len(fruits)) # number of elements
```

```python
# Modifying a list
fruits.append("strawberry")    # add at the end
fruits.remove("banana")        # remove an element
print(fruits)

fruits[0] = "grape"            # modify an element
print(fruits)
```

```python
# Looping through a list
prices = [120, 45, 89, 200, 34]

for price in prices:
    if price > 100:
        print(f"{price}€ - Expensive")
    else:
        print(f"{price}€ - Cheap")
```

```python
# Dictionaries - very important for your projects !
invoice = {
    "hotel": "Hotel Marriott",
    "date": "2024-01-15",
    "amount": 1250.50,
    "supplier": "Food & Beverage Co",
    "category": "Restaurant"
}

print(invoice["hotel"])
print(invoice["amount"])
print(invoice["date"])
```

```python
# Modifying a dictionary
invoice["amount"] = 1500.00        # modify a value
invoice["status"] = "paid"         # add a new key

print(invoice)
```

```python
# Lists of dictionaries - this is exactly your hotel project !
invoices = [
    {"supplier": "Food & Beverage Co", "amount": 1250.50, "date": "2024-01-15"},
    {"supplier": "Cleaning Services", "amount": 340.00, "date": "2024-01-16"},
    {"supplier": "Laundry Co", "amount": 89.50, "date": "2024-01-17"},
    {"supplier": "Food & Beverage Co", "amount": 2100.00, "date": "2024-01-18"},
]

total = 0
for invoice in invoices:
    total += invoice["amount"]
    print(f"{invoice['date']} - {invoice['supplier']} : {invoice['amount']}€")

print(f"\nTotal : {total}€")
```
